```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

user_stats = pd.DataFrame()

# 질문 보낸 수
user_stats['questions_sent'] = vote_point_df.groupby('user_id').size()

# 본인의 질문 중 읽힘 비율
user_stats['read_rate'] = vote_point_df.groupby('user_id')['has_read'].mean()

# 평균 응답 시간
vote_point_df['response_time_sec'] = (vote_point_df['answer_updated_at'] - vote_point_df['created_at']).dt.total_seconds()
user_stats['avg_response_time'] = vote_point_df.groupby('user_id')['response_time_sec'].mean()
```

```python
# 스케일링
scaler = StandardScaler()
scaled_data = scaler.fit_transform(user_stats)
```

```python
# elbowplot

inertias = []

for k in range(1, 16):
    model = KMeans(n_clusters=k, random_state=123)
    model.fit(scaled_data)
    inertias.append(model.inertia_)

plt.figure(figsize=(5, 5))
plt.plot(range(1, 16), inertias, marker='o')
plt.show()
```

```python
# KMeans
kmeans = KMeans(n_clusters=4, random_state=42)
user_stats['cluster'] = kmeans.fit_predict(scaled_data)
```

```python
# 각 클러스터별 평균값 비교
cluster_summary = user_stats.groupby('cluster').mean()
cluster_summary
```

```python
# 클러스터별 유저 수 계산
cluster_counts = user_stats['cluster'].value_counts().sort_index()
print(cluster_counts)
```

🔵 Cluster 0: 일반적 활동 그룹 (주류)
질문 수 많고, 읽힘률은 보통.

응답 시간도 적당 (1시간 이내).

표준적인 사용자 집단으로 추정됨.

🟢 Cluster 1: 활동 적고 반응 낮은 그룹
질문 수 적고 읽힘률 낮음.

응답도 비교적 빠른 편.

반응이 적은 사용자 or 관심이 낮은 사용자들.

🟡 Cluster 2: 반응 좋은 그룹
질문 수 중간.

읽힘률 가장 높고, 응답 시간도 길지 않음.

질문과 응답의 질적 상호작용이 좋은 사용자층.

🔴 Cluster 3: 이상치 그룹 가능성 큼
질문 수 극히 적음.

읽힘률 매우 높지만, 응답 시간 44시간 이상.
→ 서버 지연, 수동 응답, 비정상 로그 등의 가능성
→ 또는 특정 사용자 유형(예: 멘토, 관리자)이 포함됐을 수도 있음.

```python
from sklearn.metrics import silhouette_score

# X: 클러스터링에 사용한 데이터 (스케일링된 상태 추천)
# labels: 클러스터링 결과 레이블 (e.g., kmeans.labels_)

score = silhouette_score(scaled_data, kmeans.labels_)
print(f"Silhouette Score: {score:.4f}")
```

```python
# cluster 컬럼이 3이 아닌 데이터만 필터링
df_filtered = user_stats[user_stats['cluster'] != 3].copy()
```

```python
X_filtered = df_filtered[['questions_sent', 'read_rate', 'avg_response_time']]

# 스케일링 
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_filtered)

# KMeans 
kmeans = KMeans(n_clusters=3, random_state=42)
labels = kmeans.fit_predict(X_scaled)
df_filtered['new_cluster'] = labels

# 실루엣계수
score = silhouette_score(X_scaled, labels)
print(f"실루엣 점수 (k=3): {score:.4f}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 클러스터별 유저 수 계산
cluster_counts = user_stats['cluster'].value_counts().sort_index()
print(cluster_counts)

# 시각화
plt.figure(figsize=(8,5))
sns.barplot(x=cluster_counts.index, y=cluster_counts.values, palette='viridis')
plt.title('user count per cluster')
plt.xlabel('cluster')
plt.ylabel('user')
plt.show()
```

```python
# 로그 변환
df_log = np.log1p(data_numeric) 
```

```python
from scipy import stats


# z-score 계산
z_scores = np.abs(stats.zscore(df_log))

# 이상치 기준 (예: |z| > 5)
threshold = 5
outlier_mask = (z_scores > threshold)

# 이상치 개수 확인
outlier_counts = outlier_mask.sum(axis=0)
print("컬럼별 이상치 개수:")
print(df_log.columns)
print(outlier_counts)

# 이상치가 있는 행 추출
outlier_rows = df_log[(outlier_mask).any(axis=1)]
print(f"이상치 행 개수: {len(outlier_rows)}")
```

```python
# 이상치 제거: 이상치가 하나도 없는 행만 남기기
df_log_no_outliers = df_log[~(outlier_mask.any(axis=1))].copy()

print(f"제거 전 행 개수: {len(df_log)}")
print(f"제거 후 행 개수: {len(df_log_no_outliers)}")
```

```python
# Elbow Method를 사용한 클러스터 수 결정

sse = []

for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=123)
    kmeans.fit(sample_df)
    sse.append(kmeans.inertia_)

# 시각화
plt.figure(figsize=(10, 6))
plt.plot(range(1, 11), sse, marker='o')
plt.title('Elbow Method Clusters')
plt.xlabel('Clusters')
plt.ylabel('SSE')
plt.show()
```
